```python
import pandas as pd     #pandas: Used to work with tables and Excel data
import matplotlib.pyplot as plt     #matplotlib.pyplot → Used to create graphs and charts.

# Load Excel dataset

df = pd.read_excel("Student_Lifestyle_Academic_Performance_Dataset.xlsx")

print("First 10 records : ")
print(df.head(10))
```

```python
#Display Dataset Information :

print("Dataset shape : ")
print(df.shape)

print()

print("Number of Rows : ")
print(df.shape[0])

print()

print("Number of Columns : ")
print(df.shape[1])

print()

print("Columns Names : ")
print(df.columns)

print()

print("Data types : ")
print(df.dtypes)
```

```python
# Basic Student Lifestyle Analysis

print("Student Lifestyle Analysis")
print("-" * 35)

print("\nDataset Shape:")
print(f"Number of students: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")

print("\nColumn Names:")
print(df.columns.tolist())

print("\nAverage Student Statistics:")
print(df.select_dtypes(include="number").mean().round(2))

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Descriptive Statistics 

print(df.describe())
```

```python
# Missing Values

print(df.isnull())

print()

print(df.isnull().sum())


# Mean, Maximum and Minimum

columns = [
    "Study_Hours",
    "Sleep_Hours",
    "Attendance",
    "Screen_Time",
    "Exam_Score",
    "Stress_Level"
]

for column in columns:

    print("--------------------------------")

    print(column)

    print("Mean :", df[column].mean())

    print("Maximum :", df[column].max())

    print("Minimum :", df[column].min())

```

```python
# Students Scoring More Than 80

students = df[df["Exam_Score"] > 80]
print(students)


# Attendance >90 and Exam Score >75

students = df[
    (df["Attendance"] > 90) &
    (df["Exam_Score"] > 75)
]

print(students)


# Highest Exam Score

highest = df["Exam_Score"].max()

print(highest)

print()

print(df[df["Exam_Score"] == highest])

# Sorting : Descending Exam Score

print(df.sort_values(by="Exam_Score", ascending=False))

# Ascending Attendance

print(df.sort_values(by="Attendance"))


# Create Grade Column

def grade(score):

    if score >= 90:
        return "A"

    elif score >= 80:
        return "B"

    elif score >= 70:
        return "C"

    elif score >= 60:
        return "D"

    else:
        return "F"


df["Grade"] = df["Exam_Score"].apply(grade)

print(df.head())


# Count Grade Students

print(df["Grade"].value_counts())

# Mean, Median and Standard Deviation

print("Mean")

print(df["Exam_Score"].mean())

print()

print("Median")

print(df["Exam_Score"].median())

print()

print("Standard Deviation")

print(df["Exam_Score"].std())

# Correlation Matrix

correlation = df.corr(numeric_only=True)

print(correlation)


# Students Studying More Than 8 Hours but Score Below 50

students = df[
    (df["Study_Hours"] > 8) &
    (df["Exam_Score"] < 50)
]

print(students)

print()

print("Total Students")

print(len(students))
```

```python
# Bar Chart

grade_count = df["Grade"].value_counts()

plt.figure(figsize=(7,5))

plt.bar(grade_count.index, grade_count.values)

plt.title("Grade Distribution")

plt.xlabel("Grade")

plt.ylabel("Students")

plt.grid()

plt.show()
```

```python
# Pie Chart

grade_count = df["Grade"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    grade_count.values,
    labels=grade_count.index,
    autopct="%1.1f%%"
)

plt.title("Grade Distribution")

plt.show()
```

```python
# Optional Analysis (Recommended)

def screen_category(time):

    if time < 3:
        return "Low"

    elif time <= 6:
        return "Moderate"

    else:
        return "High"


df["Screen_Category"] = df["Screen_Time"].apply(screen_category)

average = df.groupby("Screen_Category")["Exam_Score"].mean()

print(average)

plt.figure(figsize=(7,5))

plt.bar(average.index, average.values)

plt.title("Average Exam Score by Screen Time Category")

plt.xlabel("Screen Time Category")

plt.ylabel("Average Exam Score")

plt.grid()

plt.show()
```
