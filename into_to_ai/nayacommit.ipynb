```python

def grade_point(mark):
    if mark >= 90:
        return 4.0
    elif mark >= 85:
        return 3.7
    elif mark >= 80:
        return 3.3
    elif mark >= 75:
        return 3.0
    elif mark >= 70:
        return 2.7
    elif mark >= 65:
        return 2.3
    elif mark >= 60:
        return 2.0
    elif mark >= 55:
        return 1.7
    elif mark >= 50:
        return 1.0
    else:
        return 0.0


marks = []

for subject in range(1, 6):
    mark = float(input(f"Enter marks for subject {subject} (0-100): "))

    if not 0 <= mark <= 100:
        raise ValueError("Marks must be between 0 and 100.")

    marks.append(mark)

grade_points = [grade_point(mark) for mark in marks]
gpa = sum(grade_points) / 5

print(f"GPA: {gpa:.2f} / 4.00")
```
